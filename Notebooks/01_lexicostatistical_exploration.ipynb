# 01 — Lexicostatistical Exploration (ODB Pipeline)
Swadesh-207 Azerbaijani–Turkish coding: cognate distribution, phonological shift breakdown, semantic dynamics.

Data source: `odb_raw_data_tables_readable.xlsx` (Tables 4.1–4.6, Appendix F).

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

XLSX = '/mnt/data/odb_raw_data_tables_readable.xlsx'
cog = pd.read_excel(XLSX, 'Table_4_2_Cognates')
phon = pd.read_excel(XLSX, 'Table_4_3_Phonology')
sem = pd.read_excel(XLSX, 'Table_4_4_Semantic')
val = pd.read_excel(XLSX, 'Table_4_5_Validation')
sample = pd.read_excel(XLSX, 'Appendix_F_Sample')

print('Total Swadesh items:', 207)
print(cog.to_string(index=False))
```

## 1. Cognate-type distribution

```python
fig, ax = plt.subplots(figsize=(8,4.5))
ax.bar(cog['Cognate type'], cog['Percentage'], color='#4C72B0')
for i,(n,p) in enumerate(zip(cog['Count'], cog['Percentage'])):
    ax.text(i, p+1, f'{n} ({p:.1f}%)', ha='center', fontsize=9)
ax.set_ylabel('% of 207 items'); ax.set_title('Cognate-type distribution (Swadesh-207, Az–Tr)')
plt.xticks(rotation=30); plt.tight_layout(); plt.show()

shared = cog.loc[cog['Cognate type'].isin(['COG-PHON','IDENT','IDENT/COG']),'Count'].sum()
print(f'Lexically shared (cognate/identical): {shared}/207 = {shared/207*100:.1f}%')
print(f"Replaced (REP-related): {cog.loc[cog['Cognate type']=='REP','Count'].iloc[0]} items")
```

## 2. Phonological shift breakdown

```python
phon2 = phon.sort_values('Count')
fig, ax = plt.subplots(figsize=(8,4.5))
ax.barh(phon2['Phonological shift'], phon2['Count'], color='#DD8452')
ax.set_xlabel('Count'); ax.set_title('Systematic sound correspondences (top shifts)')
plt.tight_layout(); plt.show()
print('Total shift-annotated items:', phon['Count'].sum())
```

## 3. Semantic dynamics

```python
sem2 = sem[sem['Semantic change']!='Total']
fig, ax = plt.subplots(figsize=(7,4.5))
ax.bar(sem2['Semantic change'], sem2['Percentage'], color='#55A868')
for i,p in enumerate(sem2['Percentage']): ax.text(i, p+1, f'{p:.2f}%', ha='center', fontsize=9)
ax.set_ylabel('% of items'); ax.set_title('Semantic-change categories')
plt.xticks(rotation=30); plt.tight_layout(); plt.show()
print('Items with no semantic change: 180/207 = 86.96%')
print('Any semantic change: 27/207 = 13.04%')
```

## 4. Validation status & sample

```python
v = val[val['Validation status']!='Total']
fig, ax = plt.subplots(figsize=(5,4))
ax.pie(v['Count'], labels=v['Validation status'], autopct='%1.1f%%', startangle=90)
ax.set_title('Coding reliability status'); plt.show()
sample
```

## Key findings
- **62.3%** of items are cognate-with-phonological-shift (COG-PHON); **19.3%** identical → lexical retention is the dominant pattern.
- Systematic correspondences: q/k (15), mək/mek (10), t/d (9), ə/e (8).
- Semantics largely stable (86.96% unchanged); replacement-driven divergence only 7.25%.
